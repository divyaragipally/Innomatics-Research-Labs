**Leap Year**

```python
def is_leap(year):
    leap = False
    
    # Write your logic here
    if year%400==0: 
        leap=True
    elif year%4==0 and year%100!=0:
        leap=True
    return leap

year = int(input())
print(is_leap(year))
```

**Python if-else**

```python
n=int(input())
if n%2!=0:
    print("Weird")
else:
    if 2<=n<=5:
        print("Not Weird")
    elif 6<=n<=20:
        print("Weird")
    else: 
        print("Not Weird")
```
