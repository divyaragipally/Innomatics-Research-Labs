```python
import pandas as pd
df = pd.read_csv("diminos_data.csv")
df.head()
```

```python
df["order_placed_at"] = pd.to_datetime(df["order_placed_at"])
df["order_delivered_at"] = pd.to_datetime(df["order_delivered_at"])
df["delivery_minutes"] = (df["order_delivered_at"] - df["order_placed_at"]).dt.total_seconds() / 60
df.head()
```

```python
df["delivery_minutes"].describe()
```

```python
delayed_orders = df[df["delivery_minutes"] > 45]
delayed_orders
delay_percentage = (len(delayed_orders) / len(df)) * 100
delay_percentage
```

```python
df["order_hour"] = df["order_placed_at"].dt.hour
orders_by_hour = df["order_hour"].value_counts().sort_index()
orders_by_hour
```

```python
orders_by_hour.idxmax(), orders_by_hour.max()
```

```python
orders_by_hour.idxmin(), orders_by_hour.min()
```

```python
avg_delivery_by_hour = df.groupby("order_hour")["delivery_minutes"].mean()
avg_delivery_by_hour
```

```python
sla_20 = (df["delivery_minutes"] <= 20).mean() * 100
sla_30 = (df["delivery_minutes"] <= 30).mean() * 100
sla_20, sla_30
```
